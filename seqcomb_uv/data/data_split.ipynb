```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.load('x.npy', allow_pickle=True)
x = x.transpose(1,0,2)
y = np.load('y.npy', allow_pickle=True)
gt = np.load('gt_exps.npy', allow_pickle=True)
gt = gt.transpose(1,0,2)

print(np.unique(y))
print(np.unique(gt))
# x.shape, y.shape, gt.shape

idx_permute = np.random.permutation(len(x))
x = x[idx_permute]
y = y[idx_permute]
gt = gt[idx_permute]
x.shape, y.shape, gt.shape
```

```python
x_train, y_train, gt_train = x[:3000], y[:3000], gt[:3000]
x_valid, y_valid, gt_valid = x[3000:5000], y[3000:5000], gt[3000:5000]
x_test, y_test, gt_test   = x[5000:], y[5000:], gt[5000:]

np.save("x_train.npy", x_train)
np.save("y_train.npy", y_train)
np.save("gt_train.npy", gt_train)

np.save("x_valid.npy", x_valid)
np.save("y_valid.npy", y_valid)
np.save("gt_valid.npy", gt_valid)

np.save("x_test.npy", x_test)
np.save("y_test.npy", y_test)
np.save("gt_test.npy", gt_test)

print(np.unique(y_test),np.unique(y_train), np.unique(y_valid))
```
